```python
def inspect_csd_file(file_path):
    print(f"--- Inspecting: {file_path} ---")
    try:
        with h5py.File(file_path, 'r') as f:
            print("Top-level keys:", list(f.keys()))
            if len(f.keys()) > 0:
                main_group_name = list(f.keys())[0]
                main_group = f[main_group_name]
                print(f"Keys inside '{main_group_name}':", list(main_group.keys()))
                if len(main_group.keys()) > 0:
                    first_sample_key = list(main_group.keys())[0]
                    first_sample_data = main_group[first_sample_key]
                    print(f"Keys inside first sample '{first_sample_key}':", list(first_sample_data.keys()))
                    if 'features' in first_sample_data.keys():
                        features_dataset = first_sample_data['features']
                        print(f"Features dataset shape for first sample: {features_dataset.shape}")
                        print(f"Features dataset dtype for first sample: {features_dataset.dtype}")
                    if 'intervals' in first_sample_data.keys():
                        intervals_dataset = first_sample_data['intervals']
                        print(f"Intervals dataset shape for first sample: {intervals_dataset.shape}")
                        print(f"Intervals dataset dtype for first sample: {intervals_dataset.dtype}")
            else:
                print("No top-level keys found.")
    except Exception as e:
        print(f"Error inspecting {file_path}: {e}")
    print("-" * (len(file_path) + 16))

# Inspect other CSD files
inspect_csd_file(os.path.join(base_path, 'acoustics', 'CMU_MOSEI_COVAREP.csd'))
inspect_csd_file(os.path.join(base_path, 'visuals', 'CMU_MOSEI_VisualFacet42.csd'))
inspect_csd_file(os.path.join(base_path, 'languages', 'CMU_MOSEI_TimestampedWordVectors.csd'))
```

```python
import torch
from torch.utils.data import Dataset
import random

class MOSEIDataset(Dataset):
    def __init__(self, dataset_h5_info, sample_ids):
        self.dataset_h5_info = dataset_h5_info
        self.sample_ids = sample_ids
        self.print_once = True

    def __len__(self):
        return len(self.sample_ids)

    def __getitem__(self, idx):
        sid = self.sample_ids[idx]

        text_data = self.dataset_h5_info['text']['file_handle'][
            self.dataset_h5_info['text']['main_group']
        ]['data'][sid]['features']

        audio_data = self.dataset_h5_info['audio']['file_handle'][
            self.dataset_h5_info['audio']['main_group']
        ]['data'][sid]['features']

        visual_data = self.dataset_h5_info['visual']['file_handle'][
            self.dataset_h5_info['visual']['main_group']
        ]['data'][sid]['features']

        label_data = self.dataset_h5_info['labels']['file_handle'][
            self.dataset_h5_info['labels']['main_group']
        ]['data'][sid]['features']

        # 🔥 DEBUG (print once)
        if self.print_once:
            print("TEXT RAW:", text_data.shape)
            print("AUDIO RAW:", audio_data.shape)
            print("VISUAL RAW:", visual_data.shape)
            print("LABEL RAW:", label_data.shape)
            self.print_once = False

        # 🔥 number of segments
        num_segments = label_data.shape[0]

        # 🔥 pick segment
        seg_idx = idx % num_segments

        # 🔥 compute segment sizes
        text_len = text_data.shape[0]
        audio_len = audio_data.shape[0]
        visual_len = visual_data.shape[0]

        t_step = text_len // num_segments
        a_step = audio_len // num_segments
        v_step = visual_len // num_segments

        # 🔥 slice SAME segment
        text = torch.tensor(
            text_data[seg_idx * t_step:(seg_idx + 1) * t_step],
            dtype=torch.float32
        )

        audio = torch.tensor(
            audio_data[seg_idx * a_step:(seg_idx + 1) * a_step],
            dtype=torch.float32
        )

        visual = torch.tensor(
            visual_data[seg_idx * v_step:(seg_idx + 1) * v_step],
            dtype=torch.float32
        )

        # 🔥 LIMIT SEQUENCE LENGTH
        MAX_TEXT = 30
        MAX_AUDIO = 100
        MAX_VISUAL = 50

        text = text[:MAX_TEXT]
        audio = audio[:MAX_AUDIO]
        visual = visual[:MAX_VISUAL]

        # 🔥 FIXED LABEL PART (CORRECTLY INDENTED)
        label_val = label_data[:].mean()

        if label_val >= 0.33:
            label = 2
        elif label_val <= -0.33:
            label = 0
        else:
            label = 1

        label = torch.tensor(label).long()

        return text, audio, visual, label
```

```python
import torch
import torch.nn as nn
import math

class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=2000):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0)
        self.register_buffer('pe', pe)

    def forward(self, x):
        return x + self.pe[:, :x.size(1)]
```

```python
class TextEncoder(nn.Module):
    def __init__(self, input_dim=300, model_dim=128, num_heads=4, num_layers=2):
        super().__init__()

        self.input_proj = nn.Linear(input_dim, model_dim)
        self.pos_enc = PositionalEncoding(model_dim)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=model_dim,
            nhead=num_heads
        )

        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers
        )

        self.attn_pool = nn.Linear(model_dim, 1)

    def forward(self, x):
        x = self.input_proj(x)
        x = self.pos_enc(x)

        # 🔥 FIX (VERY IMPORTANT)
        x = x.permute(1, 0, 2)   # (B, S, D) → (S, B, D)
        x = self.transformer(x)
        x = x.permute(1, 0, 2)   # (S, B, D) → (B, S, D)

        weights = torch.softmax(self.attn_pool(x), dim=1)
        pooled = torch.sum(weights * x, dim=1)

        return pooled
```

```python
class AudioEncoder(nn.Module):
    def __init__(self, input_dim=74, model_dim=128, num_heads=4, num_layers=2):
        super().__init__()

        self.input_proj = nn.Linear(input_dim, model_dim)
        self.pos_enc = PositionalEncoding(model_dim)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=model_dim,
            nhead=num_heads
        )

        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers
        )

        self.attn_pool = nn.Linear(model_dim, 1)

    def forward(self, x):
        x = self.input_proj(x)
        x = self.pos_enc(x)

        # 🔥 FIX for old PyTorch
        x = x.permute(1, 0, 2)
        x = self.transformer(x)
        x = x.permute(1, 0, 2)

        weights = torch.softmax(self.attn_pool(x), dim=1)
        pooled = torch.sum(weights * x, dim=1)

        return pooled
```

```python
class VisualEncoder(nn.Module):
    def __init__(self, input_dim=35, model_dim=128, num_heads=4, num_layers=2):
        super().__init__()

        self.input_proj = nn.Linear(input_dim, model_dim)
        self.pos_enc = PositionalEncoding(model_dim)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=model_dim,
            nhead=num_heads
        )

        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers
        )

        self.attn_pool = nn.Linear(model_dim, 1)

    def forward(self, x):
        x = self.input_proj(x)
        x = self.pos_enc(x)

        # 🔥 FIX for old PyTorch
        x = x.permute(1, 0, 2)
        x = self.transformer(x)
        x = x.permute(1, 0, 2)

        weights = torch.softmax(self.attn_pool(x), dim=1)
        pooled = torch.sum(weights * x, dim=1)

        return pooled
```

```python
class FusionClassifier(nn.Module):
    def __init__(self, model_dim=128, num_classes=3):  # 🔥 changed to 3
        super().__init__()

        self.cross_attn = nn.MultiheadAttention(
            embed_dim=model_dim,
            num_heads=4
        )

        self.modality_weights = nn.Parameter(torch.ones(3))

        self.fc = nn.Sequential(
            nn.Linear(model_dim * 3, 512),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(512, 256),
            nn.ReLU(),
            nn.Linear(256, num_classes)
        )

    def forward(self, text_vec, audio_vec, visual_vec):

        # 🔥 FIXED LINE
        x = torch.stack([text_vec, audio_vec, visual_vec], dim=1)

        # modality weights
        weights = torch.softmax(self.modality_weights, dim=0)
        x = x * weights.unsqueeze(0).unsqueeze(-1)

        # attention needs (seq_len, batch, dim)
        x = x.permute(1, 0, 2)

        attn_out, _ = self.cross_attn(x, x, x)

        x = x + attn_out

        # back to (B, 3, D)
        x = x.permute(1, 0, 2)

        fused = x.reshape(x.size(0), -1)

        out = self.fc(fused)
        return out
```

```python
class TriModalEmotionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.text_encoder = TextEncoder()
        self.audio_encoder = AudioEncoder()
        self.visual_encoder = VisualEncoder()
        self.fusion = FusionClassifier()

    def forward(self, text, audio, visual):

        audio = torch.nan_to_num(audio, nan=0.0, posinf=0.0, neginf=0.0)

        text_vec = self.text_encoder(text)
        audio_vec = self.audio_encoder(audio)
        visual_vec = self.visual_encoder(visual)

        output = self.fusion(text_vec, audio_vec, visual_vec)

        return output
```
